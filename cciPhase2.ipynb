# Combining the three one hour videos to create synced video

```python
import cv2
import numpy as np

# Function to resize a frame to a specific height while maintaining aspect ratio
def resize_frame(frame, height):
    (h, w) = frame.shape[:2]
    ratio = height / float(h)
    dimension = (int(w * ratio), height)
    return cv2.resize(frame, dimension, interpolation=cv2.INTER_AREA)

# Load the video files
cap1 = cv2.VideoCapture('hourlyVideos/farCam/far 00--01.mp4')
cap2 = cv2.VideoCapture('hourlyVideos/midCam/mid 00--01.mp4')
cap3 = cv2.VideoCapture('hourlyVideos/closeCam/close 00--01.mp4')

# Get properties from the first video
fps = int(cap1.get(cv2.CAP_PROP_FPS))
width = int(cap1.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap1.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the codec and create VideoWriter object
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('syncedVideo_00--01.mp4', fourcc, fps, (width * 3, height))

while True:
    ret1, frame1 = cap1.read()
    ret2, frame2 = cap2.read()
    ret3, frame3 = cap3.read()

    # Break the loop if any of the videos ends
    if not ret1 or not ret2 or not ret3:
        break

    # Resize frames to the same height
    frame2 = resize_frame(frame2, height)
    frame3 = resize_frame(frame3, height)

    # Concatenate frames horizontally
    combined_frame = np.hstack((frame1, frame2, frame3))

    # Write the combined frame to the output video
    out.write(combined_frame)

# Release everything when done
cap1.release()
cap2.release()
cap3.release()
out.release()
cv2.destroyAllWindows()
```

```python
import cv2
from matplotlib import pyplot as plt
import numpy as np
import os
 
def draw_and_show_bounding_box(image_path, box_coordinates):
    # Load the image
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert from BGR to RGB
    # Draw the bounding box
    x, y, width, height = box_coordinates
    roi = image[y:y+height, x:x+width]
    hsv_roi = cv2.cvtColor(roi, cv2.COLOR_RGB2HSV)  # Convert ROI to HSV
    cv2.rectangle(image, (x, y), (x + width, y + height), (0, 255, 0), 2)
    # Check if the ROI contains yellow
    is_yellow_present = check_for_yellow(hsv_roi)
    return is_yellow_present, image
 
def check_for_yellow(hsv_image):
    lower_yellow = np.array([15, 70, 100])  # Allowing more variation in saturation and brightness
    upper_yellow = np.array([40, 255, 255])
    # Create a mask to extract yellow areas
    mask = cv2.inRange(hsv_image, lower_yellow, upper_yellow)
    # Check if there are any yellow pixels in the ROI
    yellow_pixels = cv2.countNonZero(mask)
    return yellow_pixels > 0.2
 
def process_frames_in_folder(folder_path, box_coordinates, start_frame, end_frame, merge_threshold=10):
    yellow_frames = []  # List to store frame numbers with yellow detected
    for filename in os.listdir(folder_path):
        if filename.endswith(".jpg") or filename.endswith(".jpeg"):
            frame_number = int(os.path.splitext(filename)[0].split('_')[-1])  # Extract frame number
            
            # Only process frames within the specified range
            if start_frame <= frame_number <= end_frame:
                image_path = os.path.join(folder_path, filename)
                is_yellow_present, _ = draw_and_show_bounding_box(image_path, box_coordinates)
                
                if is_yellow_present:
                    yellow_frames.append(frame_number)
    
    yellow_frames.sort()  # Sort frame numbers in ascending order
    
    # Initialize variables to track ranges
    ranges = []
    if yellow_frames:
        start = yellow_frames[0]
        end = yellow_frames[0]
        
        # Iterate through sorted yellow_frames to find ranges
        for i in range(1, len(yellow_frames)):
            if yellow_frames[i] <= end + merge_threshold:
                end = yellow_frames[i]
            else:
                ranges.append((start, end))
                start = yellow_frames[i]
                end = yellow_frames[i]
        
        # Append the last range
        ranges.append((start, end))

    print("Detected yellow light in frame ranges:", ranges)

# Specify folder and bounding box coordinates
folder_path = "videos/29 jan/four videos/c0.mp4"
box_coordinates = (1070, 150, 10, 10)  # Adjust these coordinates as per your requirement

# Specify time range (frame numbers)
start_frame = 155050  # Change to your required start frame
end_frame = 240360     # Change to your required end frame

# Run the detection within the specified range
process_frames_in_folder(folder_path, box_coordinates, start_frame, end_frame)
```

```python
import cv2
import numpy as np

def check_for_yellow(hsv_image, pixel_threshold=50):
    """Check if the region contains enough yellow pixels."""
    lower_yellow = np.array([18, 100, 100])  # More precise lower bound
    upper_yellow = np.array([30, 255, 255])  # Reduced upper bound to avoid green

    mask = cv2.inRange(hsv_image, lower_yellow, upper_yellow)
    yellow_pixels = cv2.countNonZero(mask)

    return yellow_pixels > pixel_threshold  # Avoid false positives

def process_video(video_path, box_coordinates, merge_threshold=5, pixel_threshold=50):
    """Process the video and detect yellow lights."""
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    yellow_frames = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1

        # Extract the region of interest (ROI)
        x, y, w, h = box_coordinates
        roi = frame[y:y+h, x:x+w]

        # Convert to HSV
        hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

        # Detect yellow light
        if check_for_yellow(hsv_roi, pixel_threshold):
            yellow_frames.append(frame_count)

        # Debug: Show masked image (press 'q' to quit)
        if frame_count % 500 == 0:  # Show every 500th frame
            mask = cv2.inRange(hsv_roi, np.array([18, 100, 100]), np.array([30, 255, 255]))
            cv2.imshow("Yellow Mask", mask)
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break

    cap.release()
    cv2.destroyAllWindows()

    # Merge consecutive detections into ranges
    merged_ranges = []
    if yellow_frames:
        start, end = yellow_frames[0], yellow_frames[0]
        for i in range(1, len(yellow_frames)):
            if yellow_frames[i] <= end + merge_threshold:
                end = yellow_frames[i]
            else:
                merged_ranges.append((start, end))
                start = yellow_frames[i]
                end = yellow_frames[i]
        merged_ranges.append((start, end))

    print("Detected yellow light in frame ranges:", merged_ranges)
    return merged_ranges

# Video path and bounding box (adjust if needed)
video_path = "/home/sg1807/Desktop/sageena/cciPhase2/SLC/EB/videos/29 jan/four videos/c0.mp4"
box_coordinates = (1316, 190, 18, 18)  # Ensure this correctly isolates the traffic light

# Run detection
yellow_frame_ranges = process_video(video_path, box_coordinates)
```

```python
import cv2
import numpy as np

def check_for_green(hsv_image):
    """Check if the region of interest contains green pixels."""
    lower_green = np.array([35, 50, 50])  # More flexible lower bound
    upper_green = np.array([85, 255, 255])  # Upper bound to prevent yellow detection

    mask = cv2.inRange(hsv_image, lower_green, upper_green)
    green_pixels = cv2.countNonZero(mask)

    return green_pixels > 10  # Consider green detected if enough pixels match

def detect_green_transitions(video_path, box_coordinates):
    """Detects frames where green light stops being visible."""
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    last_green_frame = None  # Track the last frame where green was detected
    transition_frames = []  # Store frames where green disappears

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break  # Stop when the video ends

        frame_count += 1

        # Extract region of interest (ROI)
        x, y, w, h = box_coordinates
        roi = frame[y:y+h, x:x+w]

        # Convert to HSV
        hsv_roi = cv2.cvtColor(roi, cv2.COLOR_BGR2HSV)

        # Detect green light
        if check_for_green(hsv_roi):
            last_green_frame = frame_count  # Update last green frame
        elif last_green_frame is not None:
            transition_frames.append(last_green_frame)  # Store when green disappears
            last_green_frame = None  # Reset tracker

    cap.release()
    cv2.destroyAllWindows()

    if transition_frames:
        print(f"Green light disappears at frames: {transition_frames}")
    else:
        print("No green-to-non-green transitions detected.")
    
    return transition_frames

# Define video path and bounding box coordinates (x, y, width, height)
video_path = "/home/sg1807/Desktop/sageena/cciPhase2/SLC/EB/videos/29 jan/four videos/c0.mp4"  
box_coordinates = (1316, 190, 18, 18)  # Adjust based on traffic light position

# Run detection to find frames where green disappears
green_transition_frames = detect_green_transitions(video_path, box_coordinates)
```
